```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels as sm
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('dados_aula.csv')

# 1. Análise do Salário

## 1.1 Definindo Parâmetros iniciais

# tamanho da amostra
n = float(df['Salário'].count())

# calculando o número de classes
k = int(1+3.33*np.log(n))

# Calulando a amplitude total
minimo = df['Salário'].min()
maximo = df['Salário'].max()
At = maximo - minimo

# Janela de classe
h = round(At/k,2)

# Montagem dos limites de classe
limites = [minimo + i*h for i in range(k+1)]

# Categorizando os dados nas classes criadas
# right = False define que o limite esquerdo é fechado e o direito é aberto [LI,LS)
salario = df['Salário']
classes = pd.cut(salario, bins = limites, right = False, include_lowest= True)
f_i = classes.value_counts().sort_index().values

# Criando a tabela de distribuição de frequências
tab =pd.DataFrame()
tab['LI'] = limites[:-1]
tab['LS'] = limites[1:]
tab['Classes'] = [f"[{li:.2f} |- {ls:.2f})" for li,ls in zip(tab['LI'],tab['LS'])]

#frequências
tab['fi'] = f_i
tab['fri'] = tab['fi']/n
tab
```
